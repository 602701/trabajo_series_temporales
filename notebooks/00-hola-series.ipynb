```python
import pandas as pd
url = "https://raw.githubusercontent.com/Nixtla/fpp3-python/main/Assets/aus-production.csv"
df = pd.read_csv(url, sep=";")
# "1956 Q1" no es una fecha para pandas: lo convertimos a trimestre y luego a fecha
df["Quarter"] = pd.PeriodIndex(df["Quarter"].str.replace(" ", ""), freq="Q").to_timestamp()
df = df.set_index("Quarter")
df.head()
```

```python
df["Beer"].plot(title="Producción trimestral de cerveza en Australia (megalitros)");
```
